```
import random

word_list = ["python", "apple", "tiger", "laptop", "cricket"]

secret_word = random.choice(word_list)

guessed_letters = []

incorrect_guesses = 0
max_incorrect_guesses = 6

hidden_word = ["_"] * len(secret_word)

print("        WELCOME TO HANGMAN")
print("Guess the word one letter at a time.")
print(f"You have {max_incorrect_guesses} incorrect guesses allowed.\n")

while incorrect_guesses < max_incorrect_guesses:

    print("Word:", " ".join(hidden_word))
    print("Guessed Letters:", guessed_letters)
    print("Remaining Chances:", max_incorrect_guesses - incorrect_guesses)

    guess = input("\nEnter a letter: ").lower()

    if len(guess) != 1 or not guess.isalpha():
        print("Please enter only ONE alphabet letter.\n")
        continue

    if guess in guessed_letters:
        print("You already guessed that letter.\n")
        continue

    guessed_letters.append(guess)

    if guess in secret_word:

        print("Correct Guess!\n")

        for index in range(len(secret_word)):
            if secret_word[index] == guess:
                hidden_word[index] = guess

    else:
        incorrect_guesses += 1
        print("Wrong Guess!\n")

    if "_" not in hidden_word:
        print("CONGRATULATIONS! YOU WON!")
        print("The word was:", secret_word)
        break

if "_" in hidden_word:
    print("GAME OVER!")
    print("You ran out of chances.")
    print("The correct word was:", secret_word)
```
